# Fiber Placement

## Homogeneous

```python
import xara

material = xara.MultiaxialMaterial("ElasticIsotropic", E=29e3, nu=0.3)
```

```python
import veux
from xara.shapes import from_aisc

shape = from_aisc("W14x132", material=material)

veux.draw_shape(shape)
```

```python
section = xara.FrameSection("MultiaxialFiber", shape, fibers={"d": 10, "t": 1, "b": 5})

veux.draw_shape(section)
```
